###Extract Data from JSON file

####Query all JSON file from folder

```sql
%fs ls '/Volumes/gizmobox/landing/Operational_data/customers'
```

```sql
select * from json. `/Volumes/gizmobox/landing/operational_data/customers/`
```

```sql
select _metadata.file_path, _metadata.file_name, * from json. `/Volumes/gizmobox/landing/operational_data/customers/`
```

####Register files in Unity catalog using views

```sql
create or replace view 
gizmobox.bronze.v_customers as 
select  *,
_metadata.file_path, _metadata.file_name
 from json. `/Volumes/gizmobox/landing/operational_data/customers/`
```

```sql
select * from gizmobox.bronze.v_customers;
```
