###Configure access to Azure SQL Database via Lakehouse federation
####Create connection


```sql
create connection azuresql_gizmobox_con_SQL type sqlserver
options(
    host 'databricks-cert-admin.database.windows.net',
    port '1433',
    user 'vivekadmin',
    password 'Mahadev@01'
);
```

####Create foreign catalog

```sql
create foreign catalog azuresql_gizmobox_SQL_catalog
using connection azuresql_gizmobox_con_SQL
options(database 'Databricks-Certfication-SQL')
```
