###Set up the project environment for Gizmobox Data lakehouse

```sql
%fs ls 'abfss://gizmobox@dbcertificationdl.dfs.core.windows.net'
```

####Creating External location for gizmobox data lakehouse

```sql
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_Certification_ext_loc_gizmobox
    URL 'abfss://gizmobox@dbcertificationdl.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL databricks_certification_storage_cred) 
    COMMENT 'Creating external location for gizmobox data lakehouse'
    
```

```sql
%fs ls 'abfss://gizmobox@dbcertificationdl.dfs.core.windows.net/Landing'
```

```sql
Show catalogs;
```

####Creating catalog gizmobox

```sql
create catalog if not exists gizmobox
managed location 'abfss://gizmobox@dbcertificationdl.dfs.core.windows.net'
comment 'This is the container for gizmobox data lakehouse'
```

####Creating schemas
1. Landing
1. Bronze
1. Silver
1. Gold

```sql
select current_catalog();
```

```sql
use catalog gizmobox;
```

```sql
create schema if not exists gizmobox.landing
managed location 'abfss://gizmobox@dbcertificationdl.dfs.core.windows.net/Landing';
create schema if not exists gizmobox.bronze
managed location 'abfss://gizmobox@dbcertificationdl.dfs.core.windows.net/bronze';
create schema if not exists gizmobox.silver
managed location 'abfss://gizmobox@dbcertificationdl.dfs.core.windows.net/silver';
create schema if not exists gizmobox.gold
managed location 'abfss://gizmobox@dbcertificationdl.dfs.core.windows.net/gold'
```

```sql
show schemas;
```

####Create Volume

```sql
create external volume if not exists  gizmobox.landing.Operational_data
location 'abfss://gizmobox@dbcertificationdl.dfs.core.windows.net/Landing/Operational_data';
```
