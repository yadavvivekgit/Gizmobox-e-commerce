###Replace all data in a table

```sql
%sql
create schema gizmobox.demo;
```

```sql
%sql
create table gizmobox.demo.comanies
(
    company_name string,
    founded_date date,
    country string
)
```

```sql
%sql
insert into gizmobox.demo.comanies
(company_name, founded_date, country)
values
('Acme', '2010-01-01', 'USA'),
('Beta', '2012-01-01', 'Canada'),
('Gamma', '2015-01-01', 'USA'),
('Delta', '2018-01-01', 'Mexico')
```

```sql
%sql
select * from gizmobox.demo.comanies;
```

```sql
create or replace table gizmobox.demo.brz_comanies
as
select * from gizmobox.demo.comanies;
```

```sql
truncate table gizmobox.demo.brz_comanies;
```

```sql
insert overwrite table 
select * from gizmobox.demo.comanies;
```
