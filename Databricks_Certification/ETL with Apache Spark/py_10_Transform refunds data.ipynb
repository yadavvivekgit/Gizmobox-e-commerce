```python
df = spark.read.table("azuresql_gizmobox_sql_catalog.dbo.refunds")
df.display()
```

```python
from pyspark.sql.types import *
from pyspark.sql import functions as F
df_trans = df.select("refund_id",
                     "payment_id",
                     F.date_format("refund_timestamp", "yyyy-MM-dd").cast("date").alias("refund_date"),
                     F.date_format("refund_timestamp", "HH:mm:ss").alias("refund_time"),
                     "refund_amount",
                     F.split(F.col("refund_reason"), ":")[0].alias("refund_reason"),
                     F.split(F.col("refund_reason"), ":")[1].alias("refund_source")
                    )
df_trans.display()

```

```python
df_trans.writeTo("gizmobox.silver.py_refunds").createOrReplace()
```
