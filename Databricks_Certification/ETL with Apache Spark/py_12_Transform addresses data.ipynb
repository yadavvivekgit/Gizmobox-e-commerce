```python
df = spark.table("gizmobox.bronze.py_addresses")
df.display()
```

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
df_pivot_addresses = df.groupBy("customer_id")\
                       .pivot("address_type", ["billing", "shipping"])\
                       .agg(
                           max("address_line_1").alias("address_line_1"),
                           max("city").alias("city"),
                           max("state").alias("state"),
                           max("postcode").alias("postcode")
                       )
df_pivot_addresses.display()
```

```python
df_pivot_addresses.writeTo("gizmobox.silver.py_addresses").createOrReplace()
```
