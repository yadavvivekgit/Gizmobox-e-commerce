```python
df = spark.table("gizmobox.silver.py_customers")
df.display()
```

```python
df1 = spark.sql("select * from gizmobox.silver.py_addresses")
df1.display()
```

```python
df_customer_address = (df.join(df1, "customer_id", "inner").select(
                            df.customer_id,
                            df.customer_name,
                            df.date_of_birth,
                            df.email,
                            df.member_since,
                            df.telephone,
                            df1.billing_address_line_1,
                            df1.billing_city,
                            df1.billing_state,
                            df1.billing_postcode,
                            df1.shipping_address_line_1,
                            df1.shipping_city,
                            df1.shipping_state,
                            df1.shipping_postcode
                        )
)
df_customer_address.display()

```

```python
df_customer_address.writeTo("gizmobox.gold.py_customer_address").createOrReplace()
```
