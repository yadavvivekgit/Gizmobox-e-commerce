```python
df = spark.sql("select * from gizmobox.silver.py_orders")
```

```python
from pyspark.sql import functions as F
from pyspark.sql.types import *
df_agg = (df.withColumn("order_month",F.date_format("transaction_timestamp", "yyyy-MM"))
    .groupBy("customer_id", "order_month")
    .agg(
         F.countDistinct("order_id").alias("total_orders"),
         F.sum("quantity").alias("total_item_bought"),
         F.sum(F.col("price")*F.col("quantity")).alias("total_revenue")
         )
)
df_agg.display()
```

```python
df_agg.writeTo("gizmobox.gold.py_monthly_order_summary").createOrReplace()
```
