```python
from pyspark.sql.types import *
payment_schema  = StructType([
  StructField("payment_id", IntegerType(), True),
  StructField("order_id", IntegerType(), True),
  StructField("payment_timestamp", TimestampType(), True),
  StructField("payment_status", IntegerType(), True),
  StructField("payment_method", StringType(), True)
])
df = spark.read.format("csv").option("delimiter", ",").schema(payment_schema).load("abfss://gizmobox@dbcertificationdl.dfs.core.windows.net/Landing/External_data/payments/*")
df.display()
```

```python
df.writeTo("gizmobox.bronze.py_payments").createOrReplace()
```
