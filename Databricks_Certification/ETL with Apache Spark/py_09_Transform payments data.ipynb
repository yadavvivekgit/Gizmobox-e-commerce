```python
df = spark.read.table('gizmobox.bronze.py_payments')
df.display()
```

```python
from pyspark.sql.types import *
from pyspark.sql import functions as F 
df_select = df.select("payment_id",
                      "order_id",
                      F.date_format("payment_timestamp", "yyyy-MM-dd").cast("date").alias("payment_date"),
                      F.date_format("payment_timestamp", "HH:mm:ss").alias("payment_time"),
                      F.when(df.payment_status == 1, "Success")
                      .when(df.payment_status == 2, "pending")
                      .when(df.payment_status == 3, "cancelled")
                      .when(df.payment_status == 4, "Failed").alias("payment_status"),
                      "payment_method"
)
df_select.display()
```

```python
df_select.writeTo("gizmobox.silver.py_payments").createOrReplace()
```

```python
df= spark.table("gizmobox.silver.py_payments")
df.display()
```
